```python
# !pip install openpyxl # 주피터 터미널에서 설치 명령어 실행도 가능함
import openpyxl

excel_file = openpyxl.Workbook();  # 엑셀 파일 생성
excel_sheet = excel_file.active;  # 활성화된 시트 가져오기
excel_sheet.title = "크롤링데이터";  # 시트 이름 변경
excel_sheet.append(['data1', 'data2', 'data3']);  # 첫 번째 행에 헤더 추가
excel_file.save('tmp.xlsx');  # 엑셀 파일 저장
excel_file.close()
```

- 함수로 작성해보면서, 함수 작성법과 활용법 이해하기

```python
import openpyxl


def write_excel_template(filename, sheetname, listdata):
    excel_file = openpyxl.Workbook()
    excel_sheet = excel_file.active
    excel_sheet.column_dimensions['A'].width = 100  # A열 너비 조정

    if sheetname != '':
        excel_sheet.title = sheetname
    for item in listdata:
        excel_sheet.append(item)
    excel_file.save(filename)
    excel_file.close()
```

```python
import requests
from bs4 import BeautifulSoup

product_lists = list();
for page_num in range(10):
    if page_num == 0:
        res = requests.get(f'https://davelee-fun.github.io/')
    else:
        res = requests.get(f'https://davelee-fun.github.io/page={str(page_num + 1)}')
    soup = BeautifulSoup(res.content, 'html.parser')
    data = soup.select('div.card')
    for item in data:
        product_name = item.select_one('div.card-body h4.card-text')
        product_date = item.select_one('div.wrapfooter span.post-date')
        product_info = [product_name.get_text().strip(), product_date.get_text()]
        product_lists.append(product_info)

write_excel_template('tmp.xlsx', '상품정보', product_lists)
```

```python
# 엑셀 파일 읽기
excel_file = openpyxl.load_workbook('tmp.xlsx')
excel_file.sheetnames  # 시트 이름 확인, 리스트로 반환

# 시트 안에 있는 데이터 읽기
## item 에는 한 라인의 각 셀에 있는 데이터를 가져옴
## 각 데이터는 각 리스트 아이템의 value 변수로부터 실제 데이터를 가져올 수 있음
excel_sheet = excel_file["상품정보"]
print(excel_sheet)

for item in excel_sheet.rows:
    print(item[0].value, item[1].value)

excel_file.close()
```
