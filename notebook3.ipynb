### 자료구조

#### 튜플(Tuple)
* 선언 + 입력: 변수명 = (1, 2, 3, 4)
* 읽기: 변수명[index]
* 추가: 불가능
* 삭제: 불가능
* 수정: 불가능

```python
data = (1, 2, 3, 4)  # 튜플 선언
```

```python
data[1]  # 튜플 두 번째 요소 읽기
```

```python
def mul_return(a: int):
    b = a + 1
    return a, b  # 여러 값을 반환할 수 있음


result = mul_return(1)  # (1, 2), 결과 값의 타입이 Tuple
print(result)  # 출력: (1, 2)
```

```python
data1 = (1, 2, 3)
data2 = (4, 5, 6)
data3 = data1 + data2  # 튜플 합치기
print(data3)  # 출력: (1, 2, 3, 4, 5, 6)
print(data1 * 3)  # 튜플 반복
```

```python
x = 1
y = 2
x, y = y, x  # 튜플을 이용한 변수 값 교환
print(x, y)
```

```python
data1_list = list(data1)
type(data1_list);  # 리스트로 변환이 가능
```

```python
data1_tuple = tuple(data1_list)
type(data1_tuple);
print(data1_tuple) # 다시 튜플로 변환 가능
```

#### 딕셔너리 또는 사전
* 선언 + 입력: 변수명 = {} 또는 dict(), 초기값과 함께 선언, 변수명 = {키:값, 키:값}
* 읽기: 변수명[key]
* 추가: 변수명[새로운키] = 새로운 값
* 삭제: del 변수명[key]
* 수정: 변수명[key] = 새로운 값 

```python
data_dict = {'한국' : 'KR', '일본' : 'JP'} # 자바의 map과 비슷해보임
data_dict['한국'] # 읽기
data_dict['add'] = '추가'; # 추가
del data_dict['add'] # 삭제, 없는 키를 삭제할 순 없음
data_dict['일본'] = 'edit' # 수정
data_dict
```

```python
data_dict.keys(); # 키만 출력
print(data_dict.keys())
for key in data_dict.keys() :
    print(key)

data_dict.values(); # 값만 가져오기
for key in data_dict.values() :
    print(key)

print(data_dict.items());
```

#### 데이터 구조: 집합 또는 set
- 선언 + 입력: 변수명 = set(), 변수명 = {data1, data2, ...}, 변수명 = set(data1), 변수명 = set({data1, data2})
- 읽기:
- 추가:
- 삭제:
- 수정:
- 특징
    -  순서가 없음.
    - 중복 데이터 없음

```python
data_dict = dict(); # {}
data_set = set(); #  {}
# {} 이렇게 선언하면 type({}) 이게 dict (딕셔너리)로 나와서 set()을 씀.
```

```python
data_set = {'apple', 'dell', 'samsung', 'lg'}
type(data_set);  # set
```

```python
data_set = set('apple') # 데이터가 1개만 있을 때
data_set2 = set({'apple', 'dell'}) # 2개 이상
print(data_set2)
```

```python
for data in data_set:
    print(data)

print('motorola' in data_set); # False
```

```python
data1 = {'apple', 'samsung', 'lg'} # 스마트폰 생산 업체
data2 = {'samsung', 'lg', 'xaiomi'}  # TV 생산 업체
print(data1 & data2) # 교집합
print(data1 | data2) # 합집합
print(data1 - data2) # 차집합
print(data1 ^ data2) # 동일한 데이터는 제외
```

```python
# 중복 데이터 제거 방법
data_list = ['apple', 'dell', 'samsung', 'lg', 'apple']
set(data_list) # 중복 제거. 집합은 중복 데이터를 허용하지 않기 때문에 제거됨.
```
