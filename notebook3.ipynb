### 자료구조

#### 튜플(Tuple)
* 선언 + 입력: 변수명 = (1, 2, 3, 4)
* 읽기: 변수명[index]
* 추가: 불가능
* 삭제: 불가능
* 수정: 불가능

```python
data = (1, 2, 3, 4)  # 튜플 선언
```

```python
data[1]  # 튜플 두 번째 요소 읽기
```

```python
def mul_return(a: int):
    b = a + 1
    return a, b  # 여러 값을 반환할 수 있음


result = mul_return(1)  # (1, 2), 결과 값의 타입이 Tuple
print(result)  # 출력: (1, 2)
```

```python
data1 = (1, 2, 3)
data2 = (4, 5, 6)
data3 = data1 + data2  # 튜플 합치기
print(data3)  # 출력: (1, 2, 3, 4, 5, 6)
print(data1 * 3)  # 튜플 반복
```

```python
x = 1
y = 2
x, y = y, x  # 튜플을 이용한 변수 값 교환
```
