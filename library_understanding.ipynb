```python
# import 라이브러리명
# 라이브러리.함수()
# 예) statistics 라이브러리 사용하기
# - 통계 계산 기능을 제공
# - statistics 라이브러리의 mean() 함수: 평균값 계산
import statistics
data = [1, 2, 3, 4]
test = statistics.mean(data)
print(test)

# 기본 사용법2
from statistics import mean
mean(data)

from statistics import *
print(mean(data))
print(stdev(data))

# 기본 사용법3
# 라이브러리명을 줄이고 싶을 때
import statistics as stats
print(stats.mean(data))
# 라이브러리명을 줄이고 싶을 때 (다른 이름으로 변경할 때)
from statistics import mean as m
print(m(data))
```
