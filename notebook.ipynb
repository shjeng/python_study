#### 리스트 변수
1. 리스트 선언
```
    - 리스트 변수 = [] 또는 list()
```
2. 리스트 추가
```
    - 리스트 변수.append(요소) # 리스트 끝에 요소 추가
    - 리스트 변수.insert(인덱스, 요소) # 리스트 특정 위치에 요소 추가
```
3. 리스트 요소 삭제
```
    - 리스트변수.remove(요소) # 리스트에서 요소 제거, 없으면 오류
    - del 리스트변수[인덱스] # 리스트 특정 위치 요소 제거
```
4. 리스트 요소 수정
```
    - 리스트변수[인덱스] = 새로운요소 # 리스트 특정 위치 요소 수정
```

#### 18. 조건문

````python
age = input("나이는?");
age_digit = int(age);
print (age_digit);
if(age_digit >= 19): # 괄호가 없어도 if문이 작동함.
    print("당신은 성인입니다.")
else:
    print("당신은 미성년자입니다.")

if age_digit < 19 and age_digit < 40: # or: 또는, and: 그리고
    print("당신은 청년입니다.")
if not age_digit < 40:
    print('40 미만이 아니다.')

````
