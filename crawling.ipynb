```python
import requests
from bs4 import BeautifulSoup  # 라이브러리 임포트

res = requests.get('http://v.media.daum.net/v/20170615203441266')  # 웹페이지 요청
soup = BeautifulSoup(res.content, 'html.parser')  # 웹페이지 파싱
mydata = soup.find('title')  # 필요한 데이터 추출
print(mydata.get_text())  # 추출한 데이터 활용
print(mydata.string)
```

```python
html = """
<html>
    <body>
        <h1 id='title'>[1]크롤링이란?</h1>
        <p class='cssstyle'>웹페이지에서 필요한 데이터를 추출하는 것</p>
        <p id='body' align='center'>파이썬을 중심으로 다양한 웹크롤링 기술 발달</p>
    </body>
</html>
"""
soup = BeautifulSoup(html, 'html.parser')
# 태그로 검색하는 방법
data = soup.find('p')  # 첫 번째 p 태그 검색
data = soup.find('p', class_='cssstyle') # class 속성으로 검색
data = soup.find('p', 'cssstyle') # class 속성으로 검색
data = soup.find('p', attrs = {'align': 'center'}) # 속성으로 검색
data = soup.find(id='body') # id 속성으로검색
data = soup.find('p', attrs={'id': 'body', 'align': 'center'}) # 여러 속성으로 검색
print(data.string)

paragraphs = soup.find_all('p')  # 모든 p 태그 검색
for p in paragraphs:
    print(p.get_text())
```
