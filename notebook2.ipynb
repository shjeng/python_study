### 27. 함수 익히기1

https://www.inflearn.com/courses/lecture?courseId=324082&type=LECTURE&unitId=21255&tab=curriculum

##### 용어
- 인풋(input), 인자(argument, parameter): 함수에 전달되는 값

```python
def func1(parameter, parameter2):
    print("hello!", parameter, parameter2)


func1('Python', 'JAVA')  # 함수 호출, 인자 전달


def returnFuc(data1, data2):
    return data1 + data2  # 함수는 return을 통해 값을 반환할 수 있음.:


result = returnFuc('data1', 'data2')  # 함수 호출, 인자 전달
print(result)
result = returnFuc(1, 2)  # 함수 호출, 인자 전달
print(result)
```

### 28. 함수 익히기2

```python
def awe_sum(a, b):
    # 지역변수
    return a + b


a = 1
print(awe_sum(1, 2))  # 3
print(a)
```

```python
def func_wo_return(a):
    print("This is function without return for " + a + " times.")


# func_wo_return(1) # 에러 발생

# 그래서 위와 같이 사용하려면
# 아래와 같이 선언해주면 됨.
def func_wo_return(a):
    print("This is function without return for " + str(a) + " times.")


func_wo_return(1)


def func_wo_return(a):
    print("This is function without return for", a, "times.")


func_wo_return(1)
```

```python
def mul_return(a: int):
    b = a + 1
    return a, b # 여러 값을 반환할 수 있음
mul_return(1) # (1, 2)
```
