##### 용어
- 인풋(input), 인자(argument, parameter): 함수에 전달되는 값

```python
def func1(parameter, parameter2):
    print("hello!", parameter, parameter2)
func1('Python', 'JAVA')  # 함수 호출, 인자 전달
def returnFuc(data1, data2):
    return data1 + data2  # 함수는 return을 통해 값을 반환할 수 있음.:
result = returnFuc('data1', 'data2')  # 함수 호출, 인자 전달
print(result)
result = returnFuc(1, 2)  # 함수 호출, 인자 전달
print(result)
```
